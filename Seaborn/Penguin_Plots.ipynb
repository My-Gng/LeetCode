```python
import seaborn as sns
import pandas as pd
import numpy as np
```

```python
sns.get_dataset_names()
```

```python
penguins = sns.load_dataset('penguins')
penguins.head()
```

```python
penguins['species'].value_counts()
```

```python
penguins['island'].value_counts()
```

```python
sns.scatterplot(data = penguins, x = 'flipper_length_mm', y = 'body_mass_g')
```

```python
sns.scatterplot(data = penguins, x = 'flipper_length_mm', y = 'body_mass_g', hue = 'island')
```

```python
sns.scatterplot(data = penguins, x = 'flipper_length_mm', y = 'body_mass_g', hue = 'sex')
```

```python
# Styling 
# whitegrid, dark, white, ticks (sns.set_style)
# styling content
# despine() removes the top and right spines from plots
# notebook, talk, paper (sns.set_context)

sns.set_style('dark')
sns.scatterplot(data = penguins, x = 'flipper_length_mm', y = 'body_mass_g', hue = 'sex')
```

```python
sns.set_style('whitegrid')
sns.scatterplot(data = penguins, x = 'flipper_length_mm', y = 'body_mass_g', hue = 'sex')
```

```python
# strip plot
sns.stripplot(data = penguins, x = 'species', y = 'body_mass_g', hue = 'sex')
```

```python
# swarm plot
sns.set_context('talk')
sns.swarmplot(data=penguins, x='species', y='body_mass_g', hue='sex')
sns.despine()
```

```python
# histogram
sns.set_context('notebook')
sns.histplot(data=penguins, x='body_mass_g', hue='sex', multiple='stack')
```

```python
# regression plot
sns.set_context('notebook')
sns.lmplot(data=penguins, x='flipper_length_mm', y='body_mass_g')
sns.despine()
```

```python
# line plot
sns.set_context('notebook')
sns.lineplot(data=penguins, x='flipper_length_mm', y='body_mass_g', hue='species')
sns.despine()
```

```python
# joint plot
sns.jointplot(data=penguins, x='flipper_length_mm', y='body_mass_g', hue='species')
```

```python
# bar plot
sns.barplot(
    data=penguins,
    x='species',
    y='body_mass_g',
    hue='sex',
    palette=['red', 'blue']
)
```

```python
# count plot
sns.countplot(data=penguins, x='species', hue = 'island', palette=['red', 'blue'])

```

```python
# box plot
sns.boxplot(data=penguins, x='species', y='body_mass_g', hue='sex')
```

```python
# kde plot
sns.kdeplot(data=penguins, x='body_mass_g', hue='species', fill=True)
```

```python
columns = ["bill_length_mm", "bill_depth_mm", "flipper_length_mm", "body_mass_g"]
```

```python
penguins[columns].corr()
```

```python
# heatmap
# annot=True to show correlation values on the heatmap
# cmap='coolwarm' to use a diverging color palette
# center=0 to center the colormap at 0
# vmin=-1, vmax=-0.2 to set the range of the colormap
# linewidths=0.5, linecolor='black' to add lines between the cells

sns.heatmap(penguins[columns].corr(), annot=True, cmap='coolwarm', center=0, vmin=-1, vmax=-0.2, linewidths=0.5, linecolor='black')
```

```python
# pair plot
sns.pairplot(data=penguins, hue='species')
```
